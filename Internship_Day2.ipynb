```python
arr = np.array([[1,2,3], [4,5,6]])
mean = np.mean(arr)
sum_rows = np.sum(arr, axis=0)
max_cols = np.max(arr, axis=0)
variance = np.var(arr)
print(mean)
print(sum_rows)
print(max_cols)
print(variance)
```
