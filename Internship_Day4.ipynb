```python
df['Age_mean_imputed'] = df['Age'].fillna(df['Age'].mean())
df['Age_mean_imputed']
```

```python
df['Monthly_Salary_mean_imputed'] = df['Monthly_Salary'].fillna(df['Monthly_Salary'].mean())
df['Monthly_Salary_mean_imputed']
```

```python
df['Department_mode_imputed'] = df['Department'].fillna(df['Department'].mode()[0])
df['Department_mode_imputed']
```

```python
df['City_mode_imputed'] = df['City'].fillna(df['City'].mode()[0])
df['City_mode_imputed']
```

```python
df['Education_mode_imputed'] = df['Education'].fillna(df['Education'].mode()[0])
df['Education_mode_imputed']
```

```python
import pandas as pd # for data manipulation, numpy for numerical manipulation
import seaborn as sns # Data visualization
import matplotlib.pyplot as plt # Data visualization7

sns.set(style="whitegrid")

plt.figure(figsize=(10,6))

sns.boxplot(data=df, y='Monthly_Salary_mean_imputed', x='Department', palette='viridis')

plt.title('Box Plot of Monthly_Salary by Department', fontsize=16)
plt.ylabel('Monthly_Salary', fontsize=12)
plt.xlabel('Department', fontsize=12)
plt.show()
```

```python
numerical_features = df.select_dtypes(include=['float64', 'int64'])
skewness = numerical_features.skew().sort_values(ascending=False)
print("Skewness of numerical features:")
print(skewness)
```

```python
# Plot distributions of numerical features
import matplotlib.pyplot as plt
plt.figure(figsize=(15, 10))
for i, column in enumerate(numerical_features.columns, 1):
    plt.subplot((len(numerical_features.columns) // 3) + 1, 3, i)
    sns.histplot(df[column].dropna(), kde=True, bins=30, color='red')
    plt.title(f'{column} (Skewness: {skewness[column]:.2f})', fontsize=10)
    plt.xlabel(column, fontsize=8)
    plt.ylabel('Frequency', fontsize=8)

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
#scikitlearn
# Load the dataset
file_path = '/content/preprocessing_practice_dataset.csv'
titanic_data = pd.read_csv(file_path)

# Select numerical columns for scaling
numerical_columns = ['Age', 'Monthly_Salary', 'Department']

# Drop rows with missing values for simplicity
titanic_data.dropna(subset=numerical_columns, inplace=True)

# Initialize scalers
minmax_scaler = MinMaxScaler()
standard_scaler = StandardScaler()

# Apply Min-Max Scaling
titanic_data['Age_MinMax'] = minmax_scaler.fit_transform(titanic_data[['Age']])
titanic_data['Monthly_Salary_MinMax'] = minmax_scaler.fit_transform(titanic_data[['Monthly_Salary']])

# Apply Standard Scaling
titanic_data['Age_Standard'] = standard_scaler.fit_transform(titanic_data[['Age']])
titanic_data['Monthly_Salary_Standard'] = standard_scaler.fit_transform(titanic_data[['Monthly_Salary']])

# Display scaled features
titanic_data[['Age', 'Age_MinMax', 'Age_Standard','Monthly_Salary','Monthly_Salary_MinMax','Monthly_Salary_Standard']].head()
```
